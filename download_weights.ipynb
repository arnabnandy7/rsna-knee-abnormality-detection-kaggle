# Download Pretrained Weights for RSNA Knee
**Run this notebook with Internet ON, then publish /kaggle/working/weights/ as a Kaggle Dataset.**
That dataset can then be added as input to the competition notebook (Internet OFF).

```python
import timm
import torch
from pathlib import Path

SAVE_DIR = Path('/kaggle/working/weights')
SAVE_DIR.mkdir(exist_ok=True)

MODELS = [
    'efficientnet_b4',       # 19M params — direct upgrade from B2
    'efficientnet_b3',       # 12M params — intermediate step
    'convnext_tiny',         # 28M params — modern architecture
    'tf_efficientnetv2_s',   # 21M params — EfficientNetV2 Small
]

for model_name in MODELS:
    print(f'Downloading {model_name}...')
    try:
        m = timm.create_model(model_name, pretrained=True, num_classes=0)
        save_path = SAVE_DIR / f'{model_name}.pth'
        torch.save(m.state_dict(), str(save_path))
        size_mb = save_path.stat().st_size / 1e6
        print(f'  Saved {save_path.name} ({size_mb:.1f} MB)')
        del m
    except Exception as e:
        print(f'  FAILED: {e}')

print('\nAll weights saved:')
for f in sorted(SAVE_DIR.glob('*.pth')):
    print(f'  {f.name}  ({f.stat().st_size/1e6:.1f} MB)')
```

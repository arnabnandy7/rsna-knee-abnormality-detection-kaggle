# Download Pretrained EfficientNet-B2 Weights

Run this notebook with **Internet ON** to download pretrained ImageNet weights.
After it finishes, go to **Output** tab and click **New Dataset** to save the weights.

```python
import timm
import torch
import os

backbone = timm.create_model('efficientnet_b2', pretrained=True, num_classes=0)
os.makedirs('/kaggle/working/weights', exist_ok=True)
torch.save(backbone.state_dict(), '/kaggle/working/weights/efficientnet_b2.pth')
print(f"Saved backbone with {sum(p.numel() for p in backbone.parameters()):,} parameters")
print(f"Feature dim: {backbone.num_features}")
print("Done! Go to Output tab -> New Dataset to save as a Kaggle Dataset.")
```
